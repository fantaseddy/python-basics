| Type | Example |
|---|---|

| `float` | `3.14` |
| `str` | `"hello"` |
| `bool` | `True` / `False` |

Check a value's type with `type(x)`.

````python
# TODO: Predict the type of each value BEFORE running, then uncomment to check

# Integer (int)
a = 10
# Float
b = 10.5
# String (str)
c = "10"
# Boolean
d = True

print(type(a))
print(type(b))
print(type(c))
print(type(d))

# BONUS: try a + c below. Why does it break? Fix it with int(c)

# It breaks because c is declared as a string while a is declared as an integer. We are trying to perform an arithmetic operation so we cannot use string variables.
c = int(c)
a + c
````

## 2b. Type Conversion
You can convert a value from one type to another using `int()`, `float()`, `str()`, `bool()`.

```python
num = "10"
num = int(num)   # now it's an actual number, not text
```

Common use: `input()` gives you a string, so you convert it before doing math.

````python
# TODO: Fix the broken code below using type conversion

a = 10
c = "5"

# This breaks:
# result = a + c

# Fix it here:
result = a + int(c)
print(result)

# BONUS: convert 3.9 into an int. What happens to the decimal part?

b = 3.9
convert = int(b)
print(convert)

# The decimal part disappears because int does not carry decimals or fractions the same way that float does, which was what 3.9 originally was.
````

## 3. Print, Input & String Formatting
- `print()` → shows output
- `input()` → asks the user for text — **always returns a `str`**, even for numbers
- **f-strings** → the clean way to combine variables into a sentence: `f"Hi {name}"`

```python
name = input("What's your name? ")
print(f"Hello, {name}!")
```

````python
# TODO: Build a mini "About Me" generator

name = input("What's your name? ")
age = input("How old are you? ")   # this comes back as a STRING
hobby = input("What's your favorite hobby? ")

# 1. Print an f-string introduction using name, age, and hobby

print(f"Hello, I'm {name}! I have lived for {age} years. My favorite hobby is {hobby}.")

# 2. BONUS: try age + 5 directly (it breaks). Fix it with int(age) + 5

new_age = int(age) + 5
print(new_age)
````

## 4. Operators
- **Arithmetic:** `+ - * / // % **`
- **Comparison:** `== != > < >= <=` → always returns a `bool`
- **Logical:** `and`, `or`, `not` → combine conditions

````python
# TODO: Simple grade checker

score = 85
passing_score = 75

# 1. Check if the student passed (comparison operator)
passed = score >= passing_score

# 2. Check if the score is even (modulo)
is_even = score % 2 == 0

# 3. Check if the student passed AND scored above 80 ('and')
honor_roll = score >= passing_score and score > 80

print(passed, is_even, honor_roll)
````

---
## 5. Control Flow
- **Conditionals** `if / elif / else` → choose a path based on a condition
- **Loops** `for` (repeat a known number of times) / `while` (repeat until a condition changes)

```python
if score >= 75:
    print("Passed")
else:
    print("Failed")
```

````python
# TODO: Build a simple "bouncer" program

age = 17

# 1. if/elif/else:
#    "Welcome in!" if age >= 18
#    "Sorry, come back next year." if age is 17
#    "You're too young." otherwise

if age >= 18:
    print("Welcome in!")
elif age == 17:
    print("Sorry, come back next year.")
else:
    print("You're too young.")

# 2. BONUS: use a for loop to count down from 5 to 1, then print "Doors open!"

for i in range(5, 0, -1):
    print(i)

print("Doors open!")
````
